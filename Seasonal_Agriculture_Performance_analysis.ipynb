```python
# ANOVA for season test
from scipy.stats import f_oneway

seasons = df["Season"].unique()

groups = []

for season in seasons:
    groups.append(df[df["Season"] == season]["Yield_Tonnes_Ha"])

f_stat, p_value = f_oneway(*groups)

print("F-statistic:", f_stat)
print("P-value:", p_value)
```

```python
#ANOVA  result
if p_value < 0.05:
    print("There is a significant difference in yield across seasons.")
else:
    print("There is no significant difference in yield across seasons.")
```
